# Sunbrust Plot to Show Heirarchy 

```python
# import sunbrust plot on titanic Dataset 


import plotly.express as px
import seaborn as sns

# Load the Titanic dataset
titanic = sns.load_dataset('titanic')

#create a Sunbrust Plot 
fig = px.sunburst(titanic, path=['sex', 'class', 'who', 'alive', 'alone'], values='survived', color = 'sex', title='Sunburst Plot of Titanic Dataset')

# Show the plot
fig.show()
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
# Create a copy
tips_plot = tips.copy()

# Convert categorical columns to strings
tips_plot['sex'] = tips_plot['sex'].astype(str)
tips_plot['smoker'] = tips_plot['smoker'].astype(str)
tips_plot['day'] = tips_plot['day'].astype(str)
tips_plot['time'] = tips_plot['time'].astype(str)
tips_plot['tip'] = tips_plot['tip'].astype(float)

# Create Sunburst Plot
import plotly.express as px

fig = px.sunburst(
    tips_plot,
    path=['sex', 'smoker', 'day', 'time'],
    values='size', 
    color='smoker',
    title='Sunburst Plot of Tips Dataset'
)

# Show the plot
fig.show()
```
