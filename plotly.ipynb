## Plotly Data Visulization (Part I )

```python
# import Plotly library
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px 
```

```python
name = ['Jawad', 'Ali', 'Ahmed', 'Hassan', 'Hussain']
height = [5.5, 6.0, 5.8, 5.9, 6.1]

#import plotly
import plotly.express as px

# Create a plotly figure
fig = px.bar(x=name, y=height, title='Height of Individuals')

# SHow the figure
fig.show()

#SAVE Plotly figure as in png format
fig.write_image("./Outputs/height_plot.png")

# save as standlone HTML file
fig.write_html("./Outputs/height_plot.html")
```

# Plots on Titanic Data Set

```python
#import dataset
df = sns.load_dataset('titanic')

# Check the Data frame 
df.head()
```

```python
# Scatter plot using Plotly

df.dropna(inplace=True)
fig = px.scatter(df, x='age', y='fare', color='sex', size= 'age', title= 'plot of Titanic passengers age vs fare')
fig.show()

fig.write_image("./Outputs/titanic_scatter.png", scale=3) # how to save the plot in high resolution
```

```python
# Go Scatter plot using Plotly

import plotly.graph_objects as go

# create a simple line chart

fig = go.Figure(data=go.Scatter(x=[1, 2, 3, 4], y=[10, 11, 12, 13]))
fig.show()
```

```python
# Customizing the line chart with markers and line style

fig.update_layout(
    title='Customized Line Chart',
    xaxis_title=dict(text='Number'),
    yaxis_title=dict(text='Value'),
    showlegend=True
)
```

```python
# heatmap using plotly




data = np.random.rand(10, 10)

heatmap = go.Heatmap (z = data)

fig = go.Figure(data = [heatmap])

fig.show()
```

```python
# heat map corr of titanic Dataframe

fig = px.imshow(df[['age', 'fare']].corr())
fig.show()
```
