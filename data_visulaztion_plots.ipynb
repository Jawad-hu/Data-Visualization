## Data Viz 


```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# import Onlie Data from seaborn Liberary 
df = sns.load_dataset('titanic')
 
```

```python
# Now we will create a few plots usinds Pandas 
df.plot(kind= 'scatter', x = 'age', y = 'fare' , color = 'red') 
```

```python
# changing color using hex 

df.plot(kind= 'scatter', x = 'age', y = 'fare' , color = '#0f1adb') 
```

```python
df.head()
```

```python
df.groupby('sex')['fare'].mean().plot(kind = 'bar')
plt.ylabel('Average Fare')
plt.xlabel('Gender')
```

```python
# Seaborb plot 
# bar plot using seaborn on off

sns.barplot(data = df, x = 'survived', y = 'age', hue = 'class')
```

```python
# scater plot 

sns.scatterplot(data = df, y = 'fare', x = 'age', hue = 'class')
```

```python
#make scater plot via plotly
import plotly.express as px 

fig = px.scatter(df, x= 'age', y =  'fare', color = 'class')

fig.show()
```

```python
import plotly.graph_objects as go
import pandas as pd

# Read data from a csv
z_data = pd.read_csv('https://raw.githubusercontent.com/plotly/datasets/master/api_docs/mt_bruno_elevation.csv')

fig = go.Figure(data=go.Surface(z=z_data, showscale=False))
fig.update_layout(
    title=dict(text='Mt Bruno Elevation'),
    width=400, height=400,
    margin=dict(t=40, r=0, l=20, b=20)
)

name = 'default'
# Default parameters which are used when `layout.scene.camera` is not provided
camera = dict(
    up=dict(x=0, y=0, z=1),
    center=dict(x=0, y=0, z=0),
    eye=dict(x=1.25, y=1.25, z=1.25)
)

fig.update_layout(scene_camera=camera, title=name)
fig.show()
```
