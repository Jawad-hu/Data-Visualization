```python
# Data 
df = sns.load_dataset('titanic')
```

```python
df.info()
```

```python
# Scater plot
# Scater plot size 
plt.figure(figsize=(5,3))

sns.scatterplot(data = df, x='age', y='fare', hue='sex')
```

```python
# Scater plot

# Sns theme
sns.set_style('whitegrid')


# Theme 
sns.set_theme(style = 'darkgrid', font_scale = 1.5 , font = 'Times New Roman')
# Scater plot size 
plt.figure(figsize=(5,3))

#sns.scatterplot(data = df, x='age', y='fare', hue='sex')
# CHange the scatter plot color 

sns.scatterplot(data = df, x='age', y='fare', hue='sex' , palette = {'male':'green' ,'female':'red'})

# Labels
plt.xlabel('Age of Passengers')
plt.ylabel('Fare of Passengers')
plt.title('Scatter plot of Age vs Fare')

# Saving of the plot 300 dpi 
plt.savefig('./Outputs/scatter_plot.png', dpi=600)

# Saving a plot in vector format 
plt.savefig('./Outputs/scatter_plot.svg', dpi=600)
```

Interpretation:
1. we have Age of Passengers at x-axis and fare at y - axis 
2. The Scater plot shows the relationship between age and fare of titinic passengers 
3. we further grouped the passenger by their genders such as male and female (gre and red Point )
4. Two male passengers gave the highest fare in the titanic and they minght be tralvling in first class 
5. The oldest person at titanic is also a male having 80 years of age as shown in the scater plot (figure)

```python
# Boxplot

# Sns theme
sns.set_style('whitegrid')

# Theme
sns.set_theme(
    style='darkgrid',
    font_scale=1.5,
    font='Times New Roman'
)

# Boxplot size
plt.figure(figsize=(5, 3))

# Boxplot
sns.boxplot(
    data=df,
    x='sex',
    y='fare',
    palette={'male': 'green', 'female': 'red'}
)

# Labels
plt.xlabel('Gender of Passengers')
plt.ylabel('Fare of Passengers')
plt.title('Boxplot of Fare by Gender')

# changed the legend position
plt.legend(loc='upper left', title='Gender')

# Show plot
plt.show()
```

```python
sns.boxenplot(
    data=df,
    x='sex',
    y='fare',
    hue='sex',
    palette={'male': 'green', 'female': 'red'}
    
    
)

# changed  the x-tick labels
plt.xticks(ticks=[0, 1], labels=['Male', 'Female'], rotation=45)
```

```python
# Bar plot
sns.barplot(
    data=df,
    x='class',
    y='fare',
    hue='sex'
)

plt.xlabel('Passenger Class')
plt.ylabel('Average Fare')
plt.title('Average Fare by Class and Gender')

plt.show()
```
