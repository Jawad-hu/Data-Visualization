## 15 plots in Plotly 

```python
# Import libraries
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
df.info()
```

# 1 Scater Plot 


```python
fig1 = px.scatter(df, x='sepal_length', y='sepal_width', color = 'species')
fig1.show()
```

# 02 Line Plot 

```python
df_line = df.sort_values(by=['sepal_length'])
fig2 = px.line(df_line, x='sepal_length', y='sepal_width', color = 'species')
fig2.show()

fig2.write_html('./Outputs/line_plot.html', include_plotlyjs='cdn')
```

## 03 Barplot

```python
df_bar = df.groupby('species').mean().reset_index()

fig3 = px.bar(df, x='species', y='sepal_length', color = 'species')

fig3.show()
```

```python
# Seanborn does not need to make mean group 
sns.barplot(data=df, x='species', y='sepal_length', ci=None)
```

# 4 Box Plot

```python
# Boxplot

fig4 = px.box(df, x='species', y='sepal_length', color = 'species')
fig4.show()

# save the boxplot as png
fig4.write_image('./Outputs/box_plot.png', scale=2)
```

## 5 Violon Plot 


```python
# Violin plot
fig5 = px.violin(df, x='species', y='sepal_length', color = 'species', box = True)
fig5.show()
```

# Plot 6 Histogram 

```python
# histogram
fig6 = px.histogram(df, x='sepal_length', color = 'species')
fig6.show()
```

## 07 Pie Chart 

```python
# 7 Pie Chart
df_pie = df ['species'].value_counts().reset_index()
df_pie.columns = ['species', 'count']

fig7 = px.pie(df, names='species', title='Iris Species Distribution')
fig7.show()

# Save the pie chart as png
fig7.write_image('./Outputs/pie_chart.png', scale=2)
```

# 08 Scater 3D Plot 

```python
# 3D Scatter Plot
fig8 = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_length', color='species')
fig8.show()

# Save the 3D scatter plot as html
fig8.write_html('./Outputs/3d_scatter_plot.html', include_plotlyjs='cdn')
```

# 09 Area Chart 

```python
# Area Chart 

df_area = df.groupby('species')
fig9 = px.area(df, x='sepal_width', y='sepal_length', color='species')
fig9.show()
```

# 10 Buble Chart 

```python
fig10 = px.scatter(
    df,
    x='sepal_length',
    y='sepal_width',
    size='petal_length',
    color='species'
)

fig10.show()
```

## 11 Sunburst Chart 

```python
# Sunburst Chart
fig11 = px.sunburst(df, path=['species', 'sepal_length'], values='petal_length')
fig11.show()
# Save the sunburst chart as html
fig11.write_html('./Outputs/sunburst_chart.html', include_plotlyjs='cdn')
```

# 12 Parallel plot 

```python
# parallel coordinates plot
df['species_code'] = df['species'].astype('category').cat.codes

fig12 = px.parallel_coordinates(
    df,
    color='species_code',
    dimensions=[
        'sepal_length',
        'sepal_width',
        'petal_length',
        'petal_width'
    ],
    labels={
        'sepal_length': 'Sepal Length',
        'sepal_width': 'Sepal Width',
        'petal_length': 'Petal Length',
        'petal_width': 'Petal Width',
        'species_code': 'Species'
    }
)

fig12.show()

# Save the parallel coordinates plot as html
fig12.write_html('./Outputs/parallel_coordinates_plot.html', include_plotlyjs='cdn')
```

# 13 Density Contour Plot 

```python
# 13 Density Contour Plot
fig13 = px.density_contour(df, x='sepal_length', y='sepal_width', color='species')
fig13.show()

# Save the density contour plot as html
fig13.write_html('./Outputs/density_contour_plot.html', include_plotlyjs='cdn')
```

## 14 Teranery Plot 

```python
# 14 Ternary Plot
fig14 = px.scatter_ternary(df, a='sepal_length', b='sepal_width', c='petal_length', color='species')

fig14.show()
```

# 15 Polar CHart

```python
# 15 Polar Plot
df_radar = df.groupby('species').mean().reset_index()
fig15 = px.line_polar(df_radar, r='sepal_length', theta='species', line_close=True, color='species')
fig15.show()
```
